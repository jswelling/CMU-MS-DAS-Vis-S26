# Here we will mostly deal with distributions of samples, via the _displot()_ high level function.

## Load the Penguin data

```python
penguins_df = sns.load_dataset("penguins")

print(f"Data for {len(penguins_df)} specific penguins")
display(penguins_df.head())
```

# Take a moment to look at that 'multiple' option

```python
sns.histplot(data=penguins_df, x="flipper_length_mm", hue="species", multiple="stack")
```

```python
sns.histplot(data=penguins_df, x="flipper_length_mm", hue="species", multiple="layer")
```

```python
sns.histplot(data=penguins_df, x="flipper_length_mm", hue="species", multiple="dodge")
```

```python
sns.histplot(data=penguins_df, x="flipper_length_mm", hue="species", multiple="fill")
```

# histplot() is one of the low level functions called by displot()
which is short for 'distribution plot'.  The high level routine supports all these options, and it can support other idioms through _kind_ values.

Note the _rug_ values below!

```python
sns.displot(data=penguins_df, x="flipper_length_mm", hue="species", multiple="stack", kind="kde", rug=True)
```

# We can do columns.
Note that the legend is drawn in its own space!

```python
sns.displot(data=penguins_df, x="flipper_length_mm", hue="species", multiple="stack", col="island")
```

# Some other routines can let us see more clearly into the relationships:

```python
sns.jointplot(data=penguins_df, x="flipper_length_mm", y="bill_length_mm", hue="species")
```

# This one does a *scagnostic* plot (for _scatterplot diagnostics_)

```python
 sns.pairplot(data=penguins_df, hue="species")
```

# What is _kde_ anyway?

Kernel Density Estimation is a crude method of guessing the distribution that a set of samples came from.  Basically, replace
individual samples or small groups with little normal distributions and add them up.

But there is a lot of ambiguity about the _smoothing length_, the width of each little normal distribution.  For this reason,
__kde is a very approximate process__.

```python
sns.displot(data=penguins_df, x="flipper_length_mm")
```

```python
sns.displot(data=penguins_df, x="flipper_length_mm", kind="kde")
```

```python
sns.displot(data=penguins_df, x="flipper_length_mm", kind="kde", bw_adjust=0.25)
```

```python
sns.displot(data=penguins_df, x="flipper_length_mm", kind="kde", bw_adjust=2)
```

Histograms have a related problem, but at least the ambiguity is visible.

```python
sns.displot(data=penguins_df, x="flipper_length_mm", bins=20)
```
