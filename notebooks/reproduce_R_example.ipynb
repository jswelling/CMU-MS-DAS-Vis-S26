```python
%matplotlib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
```

```python
DATA_DIR_PATH = Path("/tmp/unzipped_files")
```

```python
firstgen_df = pd.read_csv( DATA_DIR_PATH / "cmu-persistence-firstgen.csv")
firstgen_df['firstgen'] = True
display(firstgen_df.dtypes)
firstgen_df.head()
```

```python
nonfirstgen_df = pd.read_csv( DATA_DIR_PATH / "cmu-persistence-nonfirstgen.csv")
nonfirstgen_df['firstgen'] = False
display(nonfirstgen_df.dtypes)
nonfirstgen_df.head()
```

```python
# We need to fix the "Initial Cohort Size" column, which didn't get converted to ints because of the commas

nonfirstgen_df["Initial Cohort Size"] = nonfirstgen_df["Initial Cohort Size"].apply(lambda s: int(s.replace(",", "")))

display(nonfirstgen_df.dtypes)
nonfirstgen_df.head()
```

```python
full_df = pd.concat([firstgen_df, nonfirstgen_df], axis=0)
full_df.head()
```

```python
sns.relplot(data=full_df, x="Cohort", y="Initial Cohort Size", hue="firstgen", kind="line", linestyle="-", marker="o")
```

* `sns.relplot` is calling `sns.lineplot` (the lower-level routine) because `kind="line"`
* `sns.lineplot` is using `plt.plot` (I think!). matplotlib is recognizing the `linestyle` and `marker` arguments.
