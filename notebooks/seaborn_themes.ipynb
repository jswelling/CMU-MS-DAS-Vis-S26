````python
%matplotlib
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
````

# Load the Penguin data

````python
penguins_df = sns.load_dataset("penguins")

print(f"Data for {len(penguins_df)} specific penguins")
display(penguins_df.head())
````

# Seaborn has a very nice styling facility
```
sns.set_theme(context="notebook", style="darkgrid", palette="deep", font="sans-serif", font_scale=1, color_codes=True, rc=None)
```
For more details, see the [aesthetics](https://seaborn.pydata.org/tutorial/aesthetics.html) and [color palette](https://seaborn.pydata.org/tutorial/color_palettes.html) tutorials.

## Draw with no theme set.  It looks very matplotlib-ish.

````python
sns.histplot(data=penguins_df, x="flipper_length_mm", hue="species", multiple="stack")
plt.show()
````

## Default Seaborn theme

context="notebook", style="darkgrid", etc

````python
sns.set_theme()

sns.histplot(data=penguins_df, x="flipper_length_mm", hue="species", multiple="stack")
plt.show()
````

## context sets sizing information, etc.

For example, posters need larger elements

````python
sns.set_theme(context="poster")

sns.histplot(data=penguins_df, x="flipper_length_mm", hue="species", multiple="stack")
plt.show()
````

## style sets the appearance of the graph frame

````python
sns.set_theme(style="white")

sns.histplot(data=penguins_df, x="flipper_length_mm", hue="species", multiple="stack")
plt.show()
````

## palette chooses a stepped color palette. They all look similar but the color intensities and contrasts change.

````python
sns.set_theme(palette="dark")

sns.histplot(data=penguins_df, x="flipper_length_mm", hue="species", multiple="stack")
plt.show()
````

## font and font_scale do just what you'd expect

Except that it's difficult to figure out the valid font names. See `matplotlib.font_manager.get_font_names()` .

````python
sns.set_theme(font="TeX Gyre Chorus", font_scale=2)

sns.histplot(data=penguins_df, x="flipper_length_mm", hue="species", multiple="stack")
plt.show()
````

## color_codes (boolean) is left as an exercise for the reader.  

## rc refers to an arbitrary rcParams structure, to override any specific setting.

````python
custom_params = {"axes.spines.right":False}
sns.set_theme(style="white", rc=custom_params)
sns.histplot(data=penguins_df, x="flipper_length_mm", hue="species", multiple="stack")
plt.show()
````

## For each category, there is a handy function to tell you with rcParam are involved.

For example, `context=` sets these:

````python
sns.plotting_context()
````

# All of these settings change the appearance of all subsequent graphs.  If you want to only change a single graph `axes_style()` and `plotting_context()` can be used as context managers.

These are the lower-level components of the high level `set_theme()` call.

````python
sns.set_theme()  # reset to defaults
sns.histplot(data=penguins_df, x="flipper_length_mm", hue="species", multiple="stack")
plt.show()
with sns.plotting_context("poster"):
    sns.histplot(data=penguins_df, x="flipper_length_mm", hue="species", multiple="stack")
    plt.show()
sns.histplot(data=penguins_df, x="flipper_length_mm", hue="species", multiple="stack")
plt.show()
````

# All of these will work for matplotlib calls too

...because Seaborn is built on top of matplotlib.

````python
with sns.axes_style("white"):
    plt.hist(penguins_df["flipper_length_mm"])
    plt.show()
````
