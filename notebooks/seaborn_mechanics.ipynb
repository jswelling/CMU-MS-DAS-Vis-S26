# Seaborn provides a number of great sample datasets
Let's pull in two:
## Penguins

```python
penguins_df = sns.load_dataset("penguins")

print(f"Data for {len(penguins_df)} specific penguins")
display(penguins_df.head())
```

* The manifold is dimension zero (you can't interpolate between penguins)
* species, island, and sex are discrete categorical values (fibers of dimension zero)
* bill length, bill depth, flipper length, and body mass are continuous real scalar values (1D fibers)

The categorical values are basically just spreadsheet entries, but there is more structure to the scalars.  If we saw the same penguin more than once we might interpolate between its two different body masses, for example.

## and a simple functional brain imaging experiment

```python
fmri_df = sns.load_dataset("fmri")

print(f"Data for {len(fmri_df.groupby(['timepoint']))} specific fmri activation curves")
print(f"with {len(fmri_df['timepoint'].unique())} time points each")
display(fmri_df.head())
```

* Several distinct subjects
* tracing signal strength over time (a 1D fiber on a 1D manifold)
