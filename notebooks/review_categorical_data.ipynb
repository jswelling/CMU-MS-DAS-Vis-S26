```python
%matplotlib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# Load the Penguin data

```python
penguins_df = sns.load_dataset("penguins")

print(f"Data for {len(penguins_df)} specific penguins")
display(penguins_df.head())
```

```python
penguins_df.dtypes
```

# Three of these columns are strings, but they come from a very restricted set.
They should be categories.


```python
penguins_df["species_ctg"] = penguins_df["species"].astype("category")
```

```python
penguins_df.dtypes
```

```python
penguins_df["species_ctg"]
```

## What actually happened here?

Our new column contains the same info as the old column, but in a new form.  Instead of storing the strings, now we just store a table of 3 strings.
Each row is an index into that table.

This greatly reduces storage, and makes statistical calculations easy!

## Floating point values can be binned to create categories

```python
print(penguins_df["bill_depth_mm"].min(), penguins_df["bill_depth_mm"].max())
```

```python
bill_min = penguins_df["bill_depth_mm"].min()
bill_max = penguins_df["bill_depth_mm"].max()
print(f"bill depth range: {bill_min} to {bill_max}")
penguins_df["bill_depth_ctg"] = pd.cut(penguins_df["bill_depth_mm"],
                                       np.linspace(bill_min, bill_max, 4),
                                       labels=["small", "medium", "large"]
                                      )
penguins_df.head()
```

```python
penguins_df.dtypes
```

```python
penguins_df["bill_depth_ctg"]
```

## Note that the bill_depth_ctg categories are *ordered*!

(see the '<' signs?)

The species_ctg values were not ordered, but we can artificially impose one.

```python
penguins_df["bill_depth_ctg"].cat.ordered
```

```python
penguins_df["species_ctg"].cat.ordered
```

# If we were to sort the records according to "species_ctg" we would get lexical ordering.

If you ask for something like `min()` of an unordered category you'll get an exception.

```python
try:
    print(penguins_df["species_ctg"].min())
except Exception as excp:
    print(f"We hit the exception: {excp}")
```

## But we can impose an arbitrary order

```python
penguins_df["species_ctg_ordered"] = penguins_df["species_ctg"].cat.set_categories(["Adelie", "Chinstrap", "Gentoo"], ordered=True)

penguins_df.head()
```

```python
penguins_df.dtypes
```

```python
penguins_df["species_ctg_ordered"]
```

```python
print(penguins_df["species_ctg_ordered"].min())
```
